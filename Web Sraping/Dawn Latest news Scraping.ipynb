```python
import requests
from bs4 import BeautifulSoup
from cffi.cffi_opcode import PRIM_FLOAT
from openpyxl.styles.builtins import title
import time
```

```python
url = requests.get("https://www.dawn.com/latest-news")
url.text
```

```python
soup = BeautifulSoup(url.text,'html.parser')
soup.prettify()
```

```python
def latest_news():
   box=soup.find_all('div',class_='sm:w-2/3 w-full sm:ml-6 sm:border-b border-gray-200')
   for boxs in box:
        classes= boxs.find('span',class_="timestamp--time timeago")
        news=boxs.find('a',class_="story__link").text
        exp=boxs.find('div',class_="story__excerpt font-georgia text-3.5 text-gray-700 overflow-hidden pb-1 px-0 sm:px-2 mt-0")
        print(f'''
        Date: {classes.text}
        Title: {news}
        Explanation: {exp.text} ''')
```

```python
if __name__=="__main__":
    while True:
      latest_news()
      time_wait=1
      print(f"Waiting for {time_wait} hour")
      time.sleep(3600)
```
