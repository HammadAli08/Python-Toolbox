```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import seaborn as sns
```

```python
data=pd.read_csv("/home/hammadali08/Vidoes/CSV file/iris.csv")
data
```

```python
x=data.iloc[:,:-1]
y=data['species']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
Rfc=RandomForestClassifier(n_estimators=100,random_state=43)
```

```python
Rfc.fit(x_train,y_train)
```

```python
Rfc.score(x_test,y_test)
```

```python
from sklearn.metrics import classification_report
```

```python
classification_report(y_test,Rfc.predict(x_test))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
Confusion_Matrix=confusion_matrix(y_test,Rfc.predict(x_test))
Confusion_Matrix
```

```python
sns.heatmap(Confusion_Matrix,annot=True)
plt.show()
```
