```python
import pandas as pd
import tensorflow
```

```python
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from keras.models import Sequential
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
cnn = Sequential()
```

```python
cnn.add(Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)))
cnn.add(MaxPooling2D(pool_size=(2, 2)))
cnn.add(Conv2D(64, (3, 3), activation='relu'))
cnn.add(MaxPooling2D(pool_size=(2, 2)))
cnn.add(Flatten())
```

```python
cnn.add(Dense(128, activation='relu'))
cnn.add(Dense(64, activation='relu'))
cnn.add(Dense(32, activation='relu'))
cnn.add(Dense(16, activation='relu'))
cnn.add(Dense(8, activation='relu'))
cnn.add(Dense(1, activation='sigmoid'))
```

```python
cnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
train_datagen = ImageDataGenerator(
        rescale=1./255,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True)

test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
        r'/home/hammadali08/Vidoes/CSV file/Photos Datset/Train data',
        target_size=(64, 64),
        batch_size=32,
        class_mode='binary')

test_generator = test_datagen.flow_from_directory(
        r'/home/hammadali08/Vidoes/CSV file/Photos Datset/test data',
        target_size=(64 , 64),
        batch_size=32,
        class_mode='binary')

cnn.fit(train_generator,
        steps_per_epoch=200,epochs=19,validation_data=test_generator)
```
