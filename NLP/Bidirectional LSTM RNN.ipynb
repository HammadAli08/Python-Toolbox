```python
import pandas as pd
dataset=pd.read_csv('/home/hammadali08/Personal/CSV file/FakeNewsNet.csv')
dataset
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemm=WordNetLemmatizer()
stop_words=set([w.lower() for w in stopwords.words('english')])

corpus=[]
for i in range(0,len(dataset)):
    review=re.sub('[^a-zA-Z]',' ',dataset['title'][i])
    review=review.lower()
    review=review.split()

    review=[lemm.lemmatize(words) for words in review if words not in stop_words]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
len(corpus)
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
vocab_size=5000
onehot_rep=[one_hot(words,vocab_size) for words in corpus]
onehot_rep
```

```python
len(onehot_rep)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
embedded_docs=pad_sequences(onehot_rep,padding='post',maxlen=20)
embedded_docs
```

```python
X=embedded_docs
y=dataset['real']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

## Model Development

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,Dense,Embedding,Bidirectional
```

```python
dimensions=60
model=Sequential()
model.add(Embedding(vocab_size,dimensions))
model.add(Bidirectional(LSTM(314)))
model.add(Dense(1,activation='sigmoid'))
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=4,batch_size=69,validation_data=(x_test,y_test))
```

```python
y_pred=model.predict(x_test)
```

```python
import numpy as np
y_pred=np.round(y_pred)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import roc_auc_score
roc_auc_score(y_test,y_pred)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
model.save('fake_news_model.keras')
```
