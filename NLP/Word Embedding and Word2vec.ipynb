# Word Embedding
It is a technique to convert words into Vectors


Words Embedding technique can be learnt in two ways:
1. Count or Frequency (BOW, OHE, TF-IDF)
2. Deep Learning Trained Models (Word2vec, Average word2vec)

## Word2Vec ==> Feature Representation
1. CBOW (Continues Bag of Words)
2. Skip Grams

## CBOW (Continues Bag of Words):
1. There will window size of the words taken in CBOW (Window size will impact the number of words that will come in one vector.
Bigger the window, Better the results)
2. In that window, the middle words is takes as an OUTPUT and other words as Context.
3. Than BOW is applied on these words.
4. ANN will be applied(Which will depends of Window Size).

## SkipGram
1. The opposite of CBOW happens while deciding Context and Output words.
2. Middle words of Window size will be considered as Context.
3. Other words will be considered as Output.

## Average Word2vec
Makes the average of all the word vectors of the sentences. And make another verctor that can go to Model Training

 ## Cosine Similarity
This is used to get the similarity between features or words with the use of Eucludian distance or manhatton distance
##### Cosine Similarity= Cos& (Cos  Angle between two words or features)
##### Distance= 1-Cosine Similarity (Distance towards ZERO means they are more similliar)

### Advantages:
1.  Vectors are of Limited Dimensions
2. Sparcity reduced(Not too many 0 and 1)
3. Semantic meaning(can know the relations of Synonyms)

# Project of Spam Filtering

```python
import pandas as pd
messages=pd.read_csv('/home/hammadali08/Downloads/SMSSpamCollection.txt',
                    sep='\t',names=["label","message"])
messages
```

```python
messages.shape
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
import re
import numpy as np
import gensim
from nltk.corpus import stopwords
lemmatizer = WordNetLemmatizer()
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z0-9]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()

    review = [lemmatizer.lemmatize(word) for word in review if word not in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
```

# Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=2500,binary=True)
x=cv.fit_transform(corpus).toarray()
x
```

```python
y=pd.get_dummies(messages['label'])
y=y.iloc[:,1]
y=y.astype('int')
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(max_depth=5)
```

```python
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report
```

```python
accuracy_score(y_test,dt.predict(x_test))
```

```python
import seaborn as sns
sns.heatmap(confusion_matrix(y_test,dt.predict(x_test)),annot=True)
```

```python
classification_report(y_test,dt.predict(x_test))
```

# Transformer:

```python
import tensorflow as tf
from tensorflow.keras import layers

# --- Multi-Head Attention Block ---
class MultiHeadSelfAttention(layers.Layer):
    def __init__(self, embed_dim, num_heads=8):
        super(MultiHeadSelfAttention, self).__init__()
        self.mha = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)

    def call(self, x):
        attn_output = self.mha(x, x)  # self-attention
        return attn_output

# --- Transformer Encoder Block ---
class TransformerBlock(layers.Layer):
    def __init__(self, embed_dim, num_heads, ff_dim, rate=0.1):
        super(TransformerBlock, self).__init__()
        self.att = MultiHeadSelfAttention(embed_dim, num_heads)
        self.ffn = tf.keras.Sequential(
            [layers.Dense(ff_dim, activation="relu"), layers.Dense(embed_dim)]
        )
        self.layernorm1 = layers.LayerNormalization(epsilon=1e-6)
        self.layernorm2 = layers.LayerNormalization(epsilon=1e-6)
        self.dropout1 = layers.Dropout(rate)
        self.dropout2 = layers.Dropout(rate)

    def call(self, inputs, training=False):
        attn_output = self.att(inputs)
        attn_output = self.dropout1(attn_output, training=training)
        out1 = self.layernorm1(inputs + attn_output)

        ffn_output = self.ffn(out1)
        ffn_output = self.dropout2(ffn_output, training=training)
        return self.layernorm2(out1 + ffn_output)

# --- Toy Example Model ---
embed_dim = 32  # embedding size
num_heads = 2   # attention heads
ff_dim = 64     # feed-forward hidden layer size
max_len = 100    # sequence length
vocab_size = 5000  # vocabulary size

inputs = layers.Input(shape=(max_len,))
embedding_layer = layers.Embedding(input_dim=vocab_size, output_dim=embed_dim)(inputs)
transformer_block = TransformerBlock(embed_dim, num_heads, ff_dim)(embedding_layer)
pooled = layers.GlobalAveragePooling1D()(transformer_block)
outputs = layers.Dense(2, activation="softmax")(pooled)  # binary classification

model = tf.keras.Model(inputs=inputs, outputs=outputs)

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=10)
```
