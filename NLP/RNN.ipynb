## What are things we have to do in Deep Learning Embedding Layer:
1. Sentences
2. Convert into OHE (With Vocabulary(Unique words) size given)
3. Padding ( Post Padding, Pre Padding)
4. OHE ==> Vectors

```python
from tensorflow.keras.preprocessing.text import one_hot
#One_Hot is for One Hot Encoding
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good']
```

```python
sent
```

```python
### Vocabulary Size: The number of Dimensions
vocab_size = 1000
vocab_size
```

My sentence will be trained on these 1000 Vocabolary Size.
### Larger Vocabulary Size means better feature Representaion.

 ## ONE HOT Representation

```python
one_hot_rep=[one_hot(words,vocab_size) for words in sent]
one_hot_rep
```

At these certain places in my Vocabulary size is ONE which represent my those words in senteneces

# Padding:
### Sentences are of different lengths, Padding makes sure to make them of equal Vectors Length by adding Zeros.

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length=7           # Every Sentence Vector will become of the length 7
embedded_sent=pad_sequences(one_hot_rep,maxlen=sent_length,padding='pre') # Post means Zeros will added at the end and Pre means zeros will added at first.
embedded_sent
```

```python
# In How many dimensions our FEATURES should be Represented
dim=12
```

```python
model=Sequential()
model.add(Embedding(vocab_size,10,input_length=sent_length))
model.compile(optimizer='adam',loss='mse')
```

```python
model.summary()
```

```python
model.predict(embedded_sent)
```
