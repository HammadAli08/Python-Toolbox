```python
import pandas as pd
dataset=pd.read_csv('/home/hammadali08/Personal/CSV file/FakeNewsNet.csv')
dataset
```

```python
dataset['real'].value_counts()
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
stopwords = set([w.lower() for w in stopwords.words('english')])
corpus=[]
for i in range(0,len(dataset)):
    review=re.sub('^[A-Za-z]',' ', dataset['title'][i])
    review=review.lower()
    review=review.split()

    review=[lemmatizer.lemmatize(words) for words in review if words not in stopwords]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
len(corpus)
```

```python
dataset['real'].shape
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
vocab_size=4000
onehot_=[one_hot(words,vocab_size) for words in corpus]
onehot_
```

```python
len(onehot_)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sent_length=20
padding=pad_sequences(onehot_,maxlen=sent_length,padding='post')
padding
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Embedding
from tensorflow.keras.layers import LSTM,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
dim=50
model=Sequential()
model.add(Embedding(input_dim=vocab_size,output_dim=dim,input_length=sent_length))
model.add(Dropout(0.2))
model.add(LSTM(400))
model.add(Dense(1,activation='sigmoid'))
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

## Training Model:

```python
X=padding
Y=dataset['real']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)
```

```python
model.fit(x_train,y_train,epochs=4,batch_size=10,validation_data=(x_test,y_test))
```

```python
import numpy as np
y_pred=model.predict(x_test)
y_pred=np.where(y_pred>0.5,1,0) # AUC ROC Curve
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
