# TF-IDF=TF*IDF
Term Frequency-Inverse Document Frequency

# Term Frequency is used for Unique words
Term Frequency=(Number of repetetions of word)/(Number of words in sentences)

# Inverse Dcument Frequency is used for words that are Common
IDF=logc (Number of Sentenes/Number of sentences containing that word)

Advatages:
1. Word Importance is getting Captured

Disadvatages:
1. Sparcity
2. Out of Vocabulary

```python
import nltk
from dateparser.parser import tokenizer
from nltk.stem import WordNetLemmatizer
Lemit=WordNetLemmatizer()
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
text='''Hero Section (Big Question Hook)

Start with a bold, conversational question like: 'Ever wondered if the news you’re reading is really true?'

Subheadline: 'Meet News Validator AI — your smart companion for fact-checking and understanding news stories.'

Add a CTA button: 'Test a News Story Now'.

Engaging Questions Section

Present a few playful, trust-building questions:

'Tired of fake news spreading everywhere?'

'Want to know the truth behind viral headlines?'

'Looking for an AI that not only checks facts but explains them too?'

Each question should smoothly lead into the feature that solves it.

How It Works

Use a 3-step flow with friendly tone:

Paste any news article or headline.

Our AI scans, validates, and checks sources.

You get a clear explanation + related trustworthy articles.

Why You’ll Love It (Features)'''
```

```python
sentence=nltk.sent_tokenize(text)
word=word_tokenize(text)
```

```python
import re
corpus=[]
for i in range(len(sentence)):
    review=re.sub('[^a-zA-Z]',' ',sentence[i])
    review=review.lower()
    review=review.split()
    review=[Lemit.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer(ngram_range=(1,2),max_features=700) # ngram basically make sure how many certain words should come in for encoding
encoded=tf.fit_transform(corpus)
```
