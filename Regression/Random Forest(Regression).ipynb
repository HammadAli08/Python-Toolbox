```python
x=data.drop('price',axis=1)
x=x.head(400)
y=data['price']
y=y.head(400)
```

```python
from sklearn.model_selection import train_test_split
```

```python
rsg=RandomForestRegressor(n_estimators=100,random_state=42)
rsg.fit(x_train,y_train)
```

```python
rsg.score(x_test,y_test)
```
